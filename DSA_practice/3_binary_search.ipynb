## Introduction to Binary Search

```python
def binary_s(arr,target):
    n = len(arr)
    left = 0
    right = n-1
    while left<=right:
        mid = (left+right)//2
        if arr[mid] == target:
            return mid
        elif target>arr[mid]:
            left = mid+1
        else:
            right = mid-1
    return -1
print(binary_s([1,4,6,8,10,23,45,67,230,],230))


## Recursive Solution

def binary_s_2(arr,target,left,right):
    n = len(arr)
    mid = (left+right)//2
    if arr[mid] == target :
        return mid
    elif target>arr[mid]:
        return binary_s_2(arr,target,mid+1,right)
    else:
        return binary_s_2(arr,target,left,mid-1)
    
    if left>right:
        return -1

print(binary_s_2([1,4,6,8,10,23,45,67,230,],230,0,len([1,4,6,8,10,23,45,67,230,])))
```

## Implementation of Lower and Upper Bound 

```python
## The lower bound of a target in a sorted array
# is the index of the first element that is greater than or equal to the target.

def lower_bnd(arr,target):
    n = len(arr)
    lower = -1
    left = 0
    right = n-1
    while left<=right:
        mid = (left+right)//2
        if arr[mid]>=target:
            lower = mid
            right = mid-1
        else:
            left = mid+1 
    return lower
print(lower_bnd([1,1,2,4,8,17,19],9))
            
 
```

```python
# The upper bound of a target in a sorted array is the index
#  of the first element that is strictly greater than the target.
def upper_bnd(arr,target):
    n = len(arr)
    upper = -1
    left = 0
    right = n-1
    while left<=right:
        mid = (left+right)//2
        if arr[mid]>target:
            upper = mid
            right = mid-1
        else:
            left = mid+1
    return upper
print(upper_bnd([1,1,2,4,8,17,19],2))
```

## Floor & Ceil in Sorted Array

```python
# The floor of a target \(x\) in a sorted array is the largest element 
# that is less than or equal to \(x\).

# The ceiling of a target \(x\) in a sorted array
# is the smallest element that is greater than or equal to \(x\).
def floor_and_ceil(arr,target):
    n = len(arr)
    left = 0
    right = n-1
    lower = -1
    upper = -1
    while left <= right:
        mid = (left+right)//2

        if arr[mid] == target:
            return [arr[mid],arr[mid]]

        elif target>=arr[mid]:
            lower = arr[mid]
            left = mid+1
        elif arr[mid]> target:
            upper = arr[mid]
            right = mid-1
        
            
    return [lower,upper]
print(floor_and_ceil([1,2,4,5,6,7,],3))
```

## Find First & Last Occurrence in Sorted Array

```python

def first_occurrence(arr, target):
    left, right = 0, len(arr) - 1
    ans = -1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            ans = mid
            right = mid - 1
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return ans


def last_occurrence(arr, target):
    left, right = 0, len(arr) - 1
    ans = -1

    while left <= right:
        mid = (left + right) // 2

        if arr[mid] == target:
            ans = mid
            left = mid + 1
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return ans


def first_last(arr, target):
    return [
        first_occurrence(arr, target),
        last_occurrence(arr, target)
    ]
```

##  Search in Rotated Sorted Array II
Problem Description
You are given an integer array nums that was originally sorted in non-decreasing order (ascending order with possible duplicates).
The array is then rotated at an unknown pivot index.

```python
def sorr_arr_2(arr,target):
    n = len(arr)
    left = 0
    right = n-1
    while left<=right:
        mid = (left+right)//2
        if arr[mid] ==  target:
            return mid
        if arr[left] == arr[right] == arr[mid]:
            left +=1
            right -=1
            continue

        if arr[mid]<=arr[right]:
            if arr[mid]<=target<=arr[right]:
                left = mid+1
            else:
                right = mid-1
        else:
            if arr[left]<=target<=arr[mid]:
                right = mid-1
            else:
                left = mid+1
    return -1
```

## Find Minimum in Rotated Sorted Array

```python
def min_sor(arr):
    n = len(arr)
    left = 0
    right = n-1
    mini = float("inf")
    while left<=right:
        mid = (left+right)//2
        if arr[mid]<=arr[right]:
            mini = min(mini,arr[mid])
            right = mid-1
        if arr[left]<= arr[right]:
            mini = min(mini,arr[left])
            left = mid+1
    return mini


        
```

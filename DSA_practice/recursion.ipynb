Implement Pow(x,n) | X raised to the power N
Problem Statement: Implement the power function pow(x, n) , which calculates the x raised to n i.e. xn.

```python
def pow(x,n):
    return x**n
print(pow(2,10))

## 
def expo(x,n):
    if n == 0 or x== 1:
        return 1 
    if n<0:
        x= 1/x
    ans = 1
    for i in range(n):
        ans *= x
    return ans
print(expo(2,10))
```

Count Good numbers
Problem Statement: A digit string is considered good if the digits at even indices (0-based) are even digits (0, 2, 4, 6, 8) and the digits at odd indices are prime digits (2, 3, 5, 7).
Given an integer n, return the total number of good digit strings of length n. As the result may be large, return it modulo 109 + 7.
A digit string is a string consisting only of the digits '0' through '9'. It may contain leading zeros.

```python
def good_num(n):
    ans = 1
    for i in range(n):
        if i % 2 == 0:
            ans  *= 5
        else:
            ans *= 4
    return ans
print(good_num(2))


```

Problem Statement

Given a positive integer n, write a recursive function to calculate the sum of all its digits.

You are not allowed to use loops (for or while).

The function should repeatedly separate the last digit and recursively process the remaining number until only one digit remains.

Examples

Example 1

Input

n = 1234

Output

10

Explanation

1 + 2 + 3 + 4 = 10

Example 2

Input

n = 7

Output

7

Example 3

Input

n = 1005

Output

6

Explanation

1 + 0 + 0 + 5 = 6
Constraints
0 ≤ n ≤ 10^18



```python
def sum_of_digits(n):
    if n < 10:
        return n

    return n % 10 + sum_of_digits(n // 10)


# Example
print(sum_of_digits(1234))
```

Problem Statement

Given a string s, determine recursively whether it is a palindrome.

A palindrome reads the same forward and backward.

You may assume:

Ignore spaces.
Ignore letter casing.
Do not use loops.

The recursive solution should compare the first and last characters and continue checking the remaining substring.

Examples

Example 1

Input

s = "madam"

Output

True

Example 2

Input

s = "RaceCar"

Output

True

Explanation

Ignore uppercase/lowercase.

Example 3

Input

s = "hello"

Output

False
Constraints
1 ≤ len(s) ≤ 1000



```python
def is_palindrome(s):
    s = s.replace(" ", "").lower()

    def helper(left, right):
        if left >= right:
            return True

        if s[left] != s[right]:
            return False

        return helper(left + 1, right - 1)

    return helper(0, len(s) - 1)


# Example
print(is_palindrome("RaceCar"))
```

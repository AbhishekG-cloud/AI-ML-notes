## Assign Cookies | Greedy Optimal O(nlogn) Solution

```python
def cookies(arr1:list,arr2:list):
    n = len(arr1)
    m = len(arr2)
    arr1.sort()
    arr2.sort()
    count = 0 
    l = 0
    r = 0
    while l < n and r < m:
        if arr2[r] >= arr1[l]:
            l+=1
            r+=1
            count+=1
        else:
            r+=1
    return count
## each sorting takes O(nlogn) 
## then total TC is O(nlogn) + O(M)


```

## Fractional Knapsack | Greedy + Sorting Optimal Solution

```python
def sack(arr:list,weight):
    n = len(arr)
    arr.sort(key=lambda x: x[0]/x[1],reverse=True)
    sum = 0
    for i in range(n):
        if weight< arr[i][1]:
            sum += (arr[i][0]/arr[i][1])*weight
            break

        else:
            sum += arr[i][0]
            weight -= arr[i][1]
    return sum
```

## Minimum Number of Coins | Greedy Optimal Solution

```python
def min_coin(arr:list,value):
    n = len(arr)
    result = []
    for i in range(n-1,-1,-1):
        while value>=arr[i]:
                value-=arr[i]
                result.append(arr[i])
    return result

```

## Lemonade Change | Greedy Solution

```python
def bills(arr):
    n = len(arr)
    my_dict = {} 

    for i in range(n):
        my_dict[arr[i]] = my_dict.get(arr[i],0)+1
        if arr[i] == 10 and my_dict["5"] > 0 :
            my_dict["5"] -= 1

        if arr[i] == 20 and my_dict["10"] > 0 and my_dict["5"] > 0:
            my_dict["5"] -= 1
            my_dict["10"] -= 1
        else:
            return False
    return True





```

## Deletion in Linked List


```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
class SingleLinkedList:
    def __init__(self):
        self.head = None

    def delete(self,val):
        if self.head.val == val:
            temp = self.head
            self.head = self.head.next
            temp.next = None
        else:
            current = self.head
            prev = None
            while current.val != val:
                prev = current
                current = current.next
            prev.next = current.next
```

## Insertion in LInked list

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
class SingleLinkedList:
    def __init__(self):
        self.head = None

    def insert(self,val,position):
        new_node = Node(val)
        if position == 0:
            new_node.next = self.head
            self.head = new_node

        else:
            current = self.head
            prev = None
            count = 0
            while current is not None and count < position:
                prev = current
                current = current.next
                count+=1
            prev.next = new_node
            new_node.next = current
```

##  Reverse a Linked List | Iterative Approach

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
class SingleLinkedList:
    def __init__(self):
        self.head = None
## Brute force: Reverse node values using a stack. The node connections remain unchanged.
    def reverse(self):
        stack = [ ]
        temp  = self.head
        while temp is not None:
            stack.append(temp.val)
            temp = temp.next
        temp = self.head
        while temp is not None:
            temp.val = stack.pop()
            temp = temp.next

## ================= Optimal Sol =======================
    def reverse(self):
        prev = None
        curr = self.head

        while curr is not None:
            next_node = curr.next  # Save next node

            curr.next = prev      # Reverse the link

            prev = curr           # Move prev forward
            curr = next_node      # Move curr forward

        self.head = prev

```

## Find Length of Loop in Linked List
 Floyd Cycle Detection Algo

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
class SingleLinkedList:
    def __init__(self):
        self.head = None
    def cycle_length(self):
        fast = self.head
        slow = self.head
    
        while fast is not None and fast.next is not None:
            slow = slow.next
            fast = fast.next.next
            
            if fast == slow:
                break
        else:
            return 0
        slow = slow.next
        count = 1
        while fast != slow:
            slow = slow.next
            count+=1
        return count 
        
```

## Odd Even Linked List
Leetcode 328:  | Rearrange Nodes

```python
class Node:
    def __init__(self,val):
        self.val = val
        self.next = None
class SingleLinkedList:
    def __init__(self):
        self.head = None

    def odd_even(self):
        if self.head is None or self.head.next is None:
            return self.head

        odd = self.head ## indexing from 1 
        even = self.head.next
        even_temp = even

        while even is not None and even.next is not None:
            odd.next = odd.next.next
            even.next = even.next.next

            odd = odd.next
            even = even.next

        odd.next = even_temp

        return self.head
```

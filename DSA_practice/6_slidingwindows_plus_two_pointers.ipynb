## Longest Substring Without Repeating Characters 

## @

```python
## ====================Brute Force=======================
def lon_str(str):
    n = len(str)
    
    ans = 1
    for i in range(n):
        count = 0
        check = set()
        for j in range(i,n):
            
            if str[j] not in check:
                count+=1
                check.add(str[j])
            else:
                break
        ans = max(ans,count)
    return ans

## =========================== Optimal ========================
def lon_str2(str):
    n = len(str)
    my_dict = dict()
    left = 0
    right = 0
    maxi = 0
    while right< n:
        if str[right] in my_dict:
            left = max(left, my_dict[str[right]]+1)
        maxi = max(maxi,right - left + 1)
        my_dict[str[right]] = right
        right+=1
    return maxi
```

## Max Consecutive Ones III | Sliding Window Approach

```python
## change from zero to one is atmost 2
## ======================= Brute =============
def cons_max_1(arr):
    n = len(arr)
    maxi = 0
    
    for i in range(n):
        zereos = 0
        for j in range(i,n):
            if arr[j]==0 :
                zereos+=1
            elif zereos > 2 :
                break
        maxi = max(maxi,j-i+1)
    return maxi
## =========================== Better ========================
def cons_max_1(arr):
    n = len(arr)
    maxi = 0
    l = 0
    r = 0
    zeros = 0

    while r<n:
        if arr[r]==0:
            zeros+=1
        while zeros>2:
            if arr[l] == 0:
                zeros-=1
            l+=1
        if zeros<=2:
            maxi = max(maxi,r-l+1)
        r+=1
    return maxi
        

## =========================== Optimal ========================
def cons_max_1(arr):
    n = len(arr)
    maxi = 0
    l = 0
    r = 0
    zeros = 0

    while r<n:
        if arr[r]==0:
            zeros+=1
        if  zeros>2:  ## only to change
            if arr[l] == 0:
                zeros-=1
            l+=1
        if zeros<=2:
            maxi = max(maxi,r-l+1)
        r+=1

```

## Fruit Into Baskets | Brute, Better & Optimal

```python
## ======================= Brute =============================
def fruits(arr):
    n =len(arr)
    max_lenght=0
    for i in range(n):
        my_set = set()
        for j in range(i,n):
            my_set.add(arr[j])
            if len(my_set)>2:
                break
            max_lenght = max(max_lenght,j-i+1)
    return max_lenght
    
## =========================== Better ========================
def fruits(arr):
    n =len(arr)
    maxi = 0
    l = 0
    r = 0
    fruit = {}
    while r<n:
        fruit[arr[r]] = fruit.get(arr[r],0)+1
        while len(fruit)>2:
            fruit[arr[l]] -= 1
            if fruit[arr[l]] == 0:
                del fruit[arr[l]]
            l+=1
        if len(fruit)<=2:
            maxi = max(maxi,r-l+1)

        r+=1
    return maxi

        

## =========================== Optimal ========================
def fruits(arr):
    n =len(arr)
    maxi = 0
    l = 0
    r = 0
    fruit = {}
    while r<n:
        fruit[arr[r]] = fruit.get(arr[r],0)+1
        if len(fruit)>2:
            fruit[arr[l]] -= 1
            if fruit[arr[l]] == 0:
                del fruit[arr[l]]
            l+=1
        if len(fruit)<=2:
            maxi = max(maxi,r-l+1)

        r+=1
    return maxi

```

## Maximum Points You Can Obtain from Cards

```python
def max_card_sum(arr):
    n = len(arr)
    left_sum = 0
    right_sum = 0
    for i in range(3):
        left_sum +=arr[i]

    maxi = left_sum
    right = n-1
    for j in range(3):
        left_sum-=arr[2-j]
        right_sum += arr[right-j]
        maxi = max(maxi,(left_sum+right_sum))
    return maxi
```

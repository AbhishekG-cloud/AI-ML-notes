```python
## Recursion
count = 0
def hello():
    global count
    
    if count > 4:
        return count
    count+=1
    print("hello")
    hello()
print(hello())
```

```python

## tail recusrion
count = 0
def name():
    global count
    if count == 4:
        return 
    count+=1
    print(count)
    name()
    print("me")
    
name()
```

```python
def n_sum(n):
    if n== 1:
        return 1
    return n + n_sum(n-1)
print(n_sum(10))
```

```python
def n_sum(sum,i,n):
    if i>n:
        print(sum)
        return
    n_sum(sum+i,i+1,n)
n_sum(0,1,10)


## TC = O(N)
## SC(stack space) = O(N)
```

```python
def fact(n):
    if n<=0 or n == 1:
        return 1
    return n*fact(n-1)
print(fact(5))
```

```python
def fact(prod,i,n):
    if i > n:
        print(prod)
        return
    fact(prod*i,i+1,n)
fact(1,1,5)
```

```python
##reverse an array 
result = []
def rev(arr:list):
    global result
    n = len(arr)
    if n==0:
        print(result)
        return
    ld = arr[-1]
    result.append(ld)
    arr.pop(-1)
    rev(arr)

rev([1,2,3,4])  

    
```

```python
## reverse  at specific position  
def rev(arr:list,left:int,right:int):
    if left >= right:
        print(arr)
        return
    arr[left],arr[right] = arr[right],arr[left]
    rev(arr,left+1,right-1)
rev([1,2,3,4,5,6],1,4) 

################################
## TC = O(N/2) sim to O(N)
## SC(stack space) = O(N/2) sim to O(N)
```

```python
## palindrome using loops
def palind(arr:list,left,right):
    n = len(arr)
    while left<right:
        if arr[left] != arr[right]:
            return False
        left +=1
        right -= 1
    return True
arr = list("asdfgfdsa")
palind(arr,0,len(arr)-1)

    
```

```python
## palindrome using recursion
def palindr(arr,left,right):
    if left>=right:
        return True
    if arr[left] != arr[right]:
        return False
    return palindr(arr,left+1,right-1)
arr = list("abcffcba")
palindr(arr,0,
        len(arr)-1)
```

```python
## fibbo number
def fibbo(n):
    if n==1 or n==2:
        return 1 
    return fibbo(n-1)+fibbo(n-2)
print(fibbo(9))
    
```
